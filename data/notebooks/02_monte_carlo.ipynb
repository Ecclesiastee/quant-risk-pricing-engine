## 02 - Simulation Monte-Carlo & Analyse de Convergence

Ce notebook explore le pricing par simulation du Mouvement Brownien Géométrique (GBM) et vérifie la loi des grands nombres via la convergence vers le prix théorique de Black-Scholes.



```python
import sys, os
import numpy as np
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath('..'))
from src.pricing.black_scholes import bs_price
from src.pricing.monte_carlo import mc_price

S, K, T, r, sigma = 100.0, 100.0, 1.0, 0.05, 0.20
bs_ref = bs_price(S, K, T, r, sigma, "call")

# Étude de la convergence selon le nombre de simulations
simulations = [500, 1000, 5000, 10000, 50000, 100000, 500000]
prices, errors, ci_lowers, ci_uppers = [], [], [], []

for n in simulations:
    p, err, (low, high) = mc_price(S, K, T, r, sigma, "call", n_sims=n, seed=42)
    prices.append(p)
    errors.append(err)
    ci_lowers.append(low)
    ci_uppers.append(high)

# Graphique de convergence
plt.figure(figsize=(10, 6))
plt.plot(simulations, prices, marker='o', label='Prix Monte-Carlo', color='blue')
plt.axhline(bs_ref, color='red', linestyle='--', label='Prix Analytique BS')
plt.fill_between(simulations, ci_lowers, ci_uppers, color='blue', alpha=0.15, label='Intervalle de Confiance 95%')

plt.xscale('log')
plt.title("Convergence de Monte-Carlo vers Black-Scholes")
plt.xlabel("Nombre de Simulations (Echelle Log)")
plt.ylabel("Prix du Call (€)")
plt.legend()
plt.show()
```
