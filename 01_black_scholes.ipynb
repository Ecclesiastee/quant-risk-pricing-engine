## 01 - Pricing Analytique Black-Scholes & Sensibilité des Greeks

Ce notebook présente le modèle de Black-Scholes (1973) pour le pricing d'options européennes et l'analyse de sensibilité des Greeks en fonction du spot et de la maturité.

```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ajout du chemin source
sys.path.append(os.path.abspath('..'))

from src.pricing.black_scholes import bs_price
from src.greeks.greeks import OptionGreeks

# Style graphique
plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
```

```python
# Paramètres de base de l'option
S_0 = 100.0   # Spot
K = 100.0     # Strike (ATM)
T = 1.0       # Maturité 1 an
r = 0.05      # Taux sans risque 5%
sigma = 0.20  # Volatilité 20%

price_call = bs_price(S_0, K, T, r, sigma, "call")
price_put = bs_price(S_0, K, T, r, sigma, "put")

print(f"--- PRICING BLACK-SCHOLES ---")
print(f"Call Price : {price_call:.4f} €")
print(f"Put Price  : {price_put:.4f} €"
```

```python
# Profil des Greeks selon la variation du prix du sous-jacent (Spot)
spots = np.linspace(50, 150, 100)

deltas_call = [OptionGreeks.delta(S, K, T, r, sigma, "call") for S in spots]
gammas = [OptionGreeks.gamma(S, K, T, r, sigma) for S in spots]
vegas = [OptionGreeks.vega(S, K, T, r, sigma) for S in spots]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(spots, deltas_call, color='blue', lw=2)
axes[0].set_title("Delta (Call)")
axes[0].set_xlabel("Spot ($S$)")
axes[0].axvline(K, color='red', linestyle='--', label='Strike K')
axes[0].legend()

axes[1].plot(spots, gammas, color='green', lw=2)
axes[1].set_title("Gamma")
axes[1].set_xlabel("Spot ($S$)")
axes[1].axvline(K, color='red', linestyle='--')

axes[2].plot(spots, vegas, color='purple', lw=2)
axes[2].set_title("Vega")
axes[2].set_xlabel("Spot ($S$)")
axes[2].axvline(K, color='red', linestyle='--')

plt.tight_layout()
```
