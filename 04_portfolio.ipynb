## 04 - Construction de Portefeuille et Agrégation des Greeks

Création d'un portefeuille hybride composé d'actions et d'options, évaluation de la valeur de marché totale et calcul des exposures globales aux Greeks.

```python
import sys, os
sys.path.append(os.path.abspath('..'))

from src.portfolio.port import Portfolio, Position

# Création d'un Portefeuille Multi-Actifs
port = Portfolio()

# Position 1: Long 100 Actions Spot = 150
port.add_position(Position(asset_type="stock", quantity=100, S=150.0))

# Position 2: Long 10 Calls AAPL K=155, T=0.5, vol=25%
port.add_position(Position(asset_type="option", quantity=10, S=150.0, K=155.0, T=0.5, r=0.03, sigma=0.25, option_type="call"))

# Position 3: Short 5 Puts AAPL K=145, T=0.5, vol=25%
port.add_position(Position(asset_type="option", quantity=-5, S=150.0, K=145.0, T=0.5, r=0.03, sigma=0.25, option_type="put"))

print(f"Valeur Totale du Portefeuille : {port.total_value():,.2f} €")

# Tableau de bord des Greeks agrégés
greeks_agg = port.aggregate_greeks()
print("\n--- RISQUES DU PORTEFEUILLE (GREEKS CONSOLIDÉS) ---")
for greek, val in greeks_agg.items():
    print(f"{greek.capitalize():10s} : {val:10.4f}")
```
